```python
# ==========================================================
# ANALISIS DE FRUTAS
# Juan Diego Sierra Sanabria
# ==========================================================

import pandas as pd

# Cargar los datos
frutas = pd.read_csv("Fruits.csv")


# Lista creada explícitamente por el estudiante
menu = [
    "Información del archivo",
    "Consultar una fruta",
    "Consultar por color",
    "Consultar por rango de precio",
    "Resumen de precios",
    "Estadísticas",
    "Ver frutas y colores disponibles",
    "Ordenar y clasificar datos",
    "Salir"
]

# Tupla creada explícitamente por el estudiante
limites_precio = (2.50, 5.00)

# Diccionario creado explícitamente por el estudiante
categorias_precio = {
    "Economica": "Precio menor a 2.50 USD",
    "Costosa": "Precio igual o mayor a 2.50 USD"
}


# ----------------------------------------------------------
# INFORMACION DEL ARCHIVO
# ----------------------------------------------------------
def mostrar_datos():
    print("\n" + "=" * 50)
    print("             INFORMACION DEL ARCHIVO")
    print("=" * 50)

    print("Cantidad de registros:", len(frutas))
    print("Cantidad de columnas:", len(frutas.columns))

    print("\nNombres de las columnas:")
    print(list(frutas.columns))

    print("\nTipos de datos:")
    print(frutas.dtypes)

    print("\nPrimeros registros:")
    print(frutas.head())


# ----------------------------------------------------------
# BUSCAR UNA FRUTA
# ----------------------------------------------------------
def consultar_fruta():
    print("\n" + "=" * 50)
    print("                 CONSULTAR FRUTA")
    print("=" * 50)

    fruta = input("Ingrese el tipo de fruta: ")

    resultado = frutas[
        frutas["type"].str.lower() == fruta.lower()
    ]

    if len(resultado) > 0:
        print("\nFruta encontrada:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo se encontró esa fruta.")


# ----------------------------------------------------------
# BUSCAR POR COLOR
# ----------------------------------------------------------
def consultar_color():
    print("\n" + "=" * 50)
    print("                  CONSULTAR COLOR")
    print("=" * 50)

    color = input("Ingrese el color: ")

    resultado = frutas[
        frutas["color"].str.lower() == color.lower()
    ]

    if len(resultado) > 0:
        print("\nRegistros encontrados:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo se encontró ese color.")


# ----------------------------------------------------------
# BUSCAR POR PRECIO
# ----------------------------------------------------------
def consultar_precio():
    print("\n" + "=" * 50)
    print("                RANGO DE PRECIOS")
    print("=" * 50)

    minimo = float(input("Precio mínimo: "))
    maximo = float(input("Precio máximo: "))

    rango = (minimo, maximo)

    resultado = frutas[
        (frutas["price usd"] >= rango[0]) &
        (frutas["price usd"] <= rango[1])
    ]

    if len(resultado) > 0:
        print("\nFrutas dentro del rango:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo existen frutas dentro de ese rango.")


# ----------------------------------------------------------
# RESUMEN DE PRECIOS
# ----------------------------------------------------------
def resumen_precios():
    print("\n" + "=" * 50)
    print("                 RESUMEN DE PRECIOS")
    print("=" * 50)

    mayor = frutas["price usd"].max()
    menor = frutas["price usd"].min()
    promedio = frutas["price usd"].mean()

    fruta_mayor = frutas[
        frutas["price usd"] == mayor
    ]

    fruta_menor = frutas[
        frutas["price usd"] == menor
    ]

    print("\nPrecio más alto:", mayor, "USD")
    print(fruta_mayor.to_string(index=False))

    print("\nPrecio más bajo:", menor, "USD")
    print(fruta_menor.to_string(index=False))

    print("\nPrecio promedio:", round(promedio, 2), "USD")


# ----------------------------------------------------------
# ESTADISTICAS
# ----------------------------------------------------------
def analizar():
    print("\n" + "=" * 50)
    print("                    ESTADISTICAS")
    print("=" * 50)

    print("1. Analizar una fruta")
    print("2. Analizar un color")
    print("3. Promedio por tipo de fruta")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":

        fruta = input("Ingrese la fruta: ")

        datos = frutas[
            frutas["type"].str.lower() == fruta.lower()
        ]

        if len(datos) > 0:
            print("\nEstadísticas de la fruta:")
            print(datos["price usd"].describe())
        else:
            print("\nNo se encontró esa fruta.")

    elif opcion == "2":

        color = input("Ingrese el color: ")

        datos = frutas[
            frutas["color"].str.lower() == color.lower()
        ]

        if len(datos) > 0:
            print("\nEstadísticas del color:")
            print(datos["price usd"].describe())
        else:
            print("\nNo se encontró ese color.")

    elif opcion == "3":

        promedio_frutas = (
            frutas.groupby("type")["price usd"]
            .mean()
            .sort_values()
        )

        print("\nPrecio promedio por tipo de fruta:")
        print(promedio_frutas)

    else:
        print("\nOpción no válida.")


# ----------------------------------------------------------
# MOSTRAR TIPOS Y COLORES
# ----------------------------------------------------------
def mostrar_opciones():
    print("\n" + "=" * 50)
    print("                DATOS DISPONIBLES")
    print("=" * 50)

    print("\nTipos de fruta:")
    print(sorted(frutas["type"].unique()))

    print("\nColores:")
    print(sorted(frutas["color"].unique()))


# ----------------------------------------------------------
# ORDENAR Y CLASIFICAR
# ----------------------------------------------------------
def organizar_datos():

    print("\n" + "=" * 50)
    print("              ORGANIZAR DATOS")
    print("=" * 50)

    # Lista para guardar las categorías
    clasificacion = []

    # Clasificación mediante un ciclo
    for precio in frutas["price usd"]:

        if precio < limites_precio[0]:
            clasificacion.append("Economica")
        else:
            clasificacion.append("Costosa")

    # Crear una nueva columna
    frutas["categoria"] = clasificacion

    print("\nDatos ordenados por precio:")

    ordenados = frutas.sort_values(
        "price usd",
        ascending=True
    )

    print(ordenados.to_string(index=False))

    print("\nNueva columna 'categoria':")

    print(
        frutas[
            ["type", "price usd", "categoria"]
        ].head(10).to_string(index=False)
    )

    print("\nCategorías utilizadas:")
    for categoria, descripcion in categorias_precio.items():
        print("-", categoria, ":", descripcion)


# ----------------------------------------------------------
# MENU PRINCIPAL
# ----------------------------------------------------------

while True:

    print("\n\n" + "=" * 55)
    print("             ANALISIS DE FRUTAS")
    print("=" * 55)

    for numero, opcion in enumerate(menu, start=1):
        print(f"{numero}. {opcion}")

    seleccion = input("\nSeleccione una opción: ")

    if seleccion == "1":
        mostrar_datos()

    elif seleccion == "2":
        consultar_fruta()

    elif seleccion == "3":
        consultar_color()

    elif seleccion == "4":
        consultar_precio()

    elif seleccion == "5":
        resumen_precios()

    elif seleccion == "6":
        analizar()

    elif seleccion == "7":
        mostrar_opciones()

    elif seleccion == "8":
        organizar_datos()

    elif seleccion == "9":
        print("\n" + "=" * 55)
        print("Gracias por utilizar el programa.")
        print("=" * 55)
        break

    else:
        print("\nOpción no válida. Intente nuevamente.")
```
