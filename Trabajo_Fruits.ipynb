# Análisis y consulta de datos de frutas

Este programa permite cargar y analizar información almacenada en el archivo `Fruits.csv` utilizando la biblioteca pandas de Python.

El programa permite consultar frutas y colores, filtrar precios, calcular estadísticas, identificar valores diferentes, ordenar los datos y crear una nueva categoría de precios.

## 1. Carga del archivo

En este bloque se importa la biblioteca pandas, que permite trabajar con datos organizados en filas y columnas.

Posteriormente, se utiliza `read_csv()` para cargar el archivo `Fruits.csv` y almacenarlo en un DataFrame llamado `frutas`.

Un DataFrame es una estructura de pandas que permite organizar y manipular datos en forma de tabla.

```python
import pandas as pd

frutas = pd.read_csv("Fruits.csv")
```

## 2. Estructuras de datos utilizadas

En este bloque se crean tres estructuras de datos solicitadas en el ejercicio.

La primera es una lista llamada `menu`, que contiene las diferentes opciones del programa.

La segunda es una tupla llamada `limites_precio`, que contiene los valores utilizados como referencia para clasificar los precios.

La tercera es un diccionario llamado `categorias_precio`, que relaciona cada categoría con una descripción.

```python
# Lista creada explícitamente
menu = [
    "Información del archivo",
    "Consultar una fruta",
    "Consultar por color",
    "Consultar por rango de precio",
    "Resumen de precios",
    "Estadísticas",
    "Ver frutas y colores disponibles",
    "Ordenar y clasificar datos",
    "Salir"
]

# Tupla creada explícitamente
limites_precio = (2.50, 5.00)

# Diccionario creado explícitamente
categorias_precio = {
    "Economica": "Precio menor a 2.50 USD",
    "Costosa": "Precio igual o mayor a 2.50 USD"
}
```

## 3. Información general del archivo

La función `mostrar_datos()` permite conocer las características básicas del conjunto de datos.

Se muestra la cantidad de registros, la cantidad de columnas, los nombres de las columnas, los tipos de datos y los primeros registros del DataFrame.

Esto permite verificar cómo está organizado el archivo antes de realizar los diferentes análisis.

```python
def mostrar_datos():
    print("\n" + "=" * 50)
    print("             INFORMACION DEL ARCHIVO")
    print("=" * 50)

    print("Cantidad de registros:", len(frutas))
    print("Cantidad de columnas:", len(frutas.columns))

    print("\nNombres de las columnas:")
    print(list(frutas.columns))

    print("\nTipos de datos:")
    print(frutas.dtypes)

    print("\nPrimeros registros:")
    print(frutas.head())
```

## 4. Consulta de una fruta

La función `consultar_fruta()` permite buscar una fruta específica dentro del DataFrame.

El usuario introduce el nombre de la fruta y el programa compara este valor con la columna `type`.

Se utiliza `lower()` para convertir los textos a minúsculas y evitar problemas cuando el usuario escribe una fruta utilizando mayúsculas o minúsculas.

Finalmente, mediante una condición se determina si se encontraron registros.

```python
def consultar_fruta():
    fruta = input("Ingrese el tipo de fruta: ")

    resultado = frutas[
        frutas["type"].str.lower() == fruta.lower()
    ]

    if len(resultado) > 0:
        print("\nFruta encontrada:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo se encontró esa fruta.")
```

## 5. Consulta por color

La función `consultar_color()` permite buscar todos los registros que tengan un color determinado.

El usuario introduce el color y el programa compara el valor ingresado con la columna `color`.

Si existen registros que coincidan, se muestran en pantalla. Si no se encuentra ningún registro, se muestra un mensaje indicando que el color no fue encontrado.

```python
def consultar_color():
    color = input("Ingrese el color: ")

    resultado = frutas[
        frutas["color"].str.lower() == color.lower()
    ]

    if len(resultado) > 0:
        print("\nRegistros encontrados:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo se encontró ese color.")
```

## 6. Consulta por rango de precio

La función `consultar_precio()` permite seleccionar un rango de precios.

El usuario introduce un precio mínimo y un precio máximo. Estos valores se almacenan en una tupla llamada `rango`.

Después se aplican dos condiciones:

- El precio debe ser mayor o igual al mínimo.
- El precio debe ser menor o igual al máximo.

El operador `&` permite combinar ambas condiciones.

```python
def consultar_precio():
    minimo = float(input("Precio mínimo: "))
    maximo = float(input("Precio máximo: "))

    rango = (minimo, maximo)

    resultado = frutas[
        (frutas["price usd"] >= rango[0]) &
        (frutas["price usd"] <= rango[1])
    ]

    if len(resultado) > 0:
        print("\nFrutas dentro del rango:")
        print(resultado.to_string(index=False))
    else:
        print("\nNo existen frutas dentro de ese rango.")
```

## 7. Análisis general de los precios

La función `resumen_precios()` permite analizar los precios de todas las frutas.

Se utiliza:

- `max()` para encontrar el precio más alto.
- `min()` para encontrar el precio más bajo.
- `mean()` para calcular el precio promedio.

Después se filtran los registros correspondientes al precio máximo y mínimo para identificar las frutas asociadas a estos valores.

```python
def resumen_precios():
    mayor = frutas["price usd"].max()
    menor = frutas["price usd"].min()
    promedio = frutas["price usd"].mean()

    fruta_mayor = frutas[
        frutas["price usd"] == mayor
    ]

    fruta_menor = frutas[
        frutas["price usd"] == menor
    ]

    print("\nPrecio más alto:", mayor, "USD")
    print(fruta_mayor.to_string(index=False))

    print("\nPrecio más bajo:", menor, "USD")
    print(fruta_menor.to_string(index=False))

    print("\nPrecio promedio:", round(promedio, 2), "USD")
```

## 8. Análisis estadístico

La función `analizar()` permite realizar diferentes análisis estadísticos.

El usuario puede seleccionar una fruta, un color o calcular el precio promedio de cada tipo de fruta.

Para analizar una fruta o un color se utiliza `describe()`, que proporciona información estadística como la cantidad de datos, promedio, desviación estándar, mínimo, cuartiles y máximo.

También se utiliza `groupby()` para agrupar los registros según el tipo de fruta y posteriormente calcular el precio promedio de cada grupo.

```python
def analizar():
    print("1. Analizar una fruta")
    print("2. Analizar un color")
    print("3. Promedio por tipo de fruta")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        fruta = input("Ingrese la fruta: ")

        datos = frutas[
            frutas["type"].str.lower() == fruta.lower()
        ]

        if len(datos) > 0:
            print("\nEstadísticas de la fruta:")
            print(datos["price usd"].describe())
        else:
            print("\nNo se encontró esa fruta.")

    elif opcion == "2":
        color = input("Ingrese el color: ")

        datos = frutas[
            frutas["color"].str.lower() == color.lower()
        ]

        if len(datos) > 0:
            print("\nEstadísticas del color:")
            print(datos["price usd"].describe())
        else:
            print("\nNo se encontró ese color.")

    elif opcion == "3":
        promedio_frutas = (
            frutas.groupby("type")["price usd"]
            .mean()
            .sort_values()
        )

        print("\nPrecio promedio por tipo de fruta:")
        print(promedio_frutas)

    else:
        print("\nOpción no válida.")
```

## 9. Valores disponibles en el conjunto de datos

La función `mostrar_opciones()` permite identificar los diferentes tipos de frutas y colores existentes en el archivo.

Para esto se utiliza `unique()`, que permite obtener los valores diferentes de una columna.

También se utiliza `sorted()` para ordenar los valores obtenidos y presentarlos de una manera más organizada.

```python
def mostrar_opciones():
    print("\nTipos de fruta:")
    print(sorted(frutas["type"].unique()))

    print("\nColores:")
    print(sorted(frutas["color"].unique()))
```

## 10. Clasificación y ordenamiento de los datos

La función `organizar_datos()` utiliza un ciclo `for` para recorrer todos los precios del DataFrame.

Cada precio se compara con el límite establecido en la tupla `limites_precio`.

Si el precio es menor a 2.50 USD, se clasifica como `Economica`. En caso contrario, se clasifica como `Costosa`.

Las clasificaciones se almacenan en una lista y posteriormente se utilizan para crear una nueva columna llamada `categoria`.

Finalmente, los registros se ordenan de menor a mayor precio utilizando `sort_values()`.

```python
def organizar_datos():
    clasificacion = []

    for precio in frutas["price usd"]:
        if precio < limites_precio[0]:
            clasificacion.append("Economica")
        else:
            clasificacion.append("Costosa")

    frutas["categoria"] = clasificacion

    print("\nDatos ordenados por precio:")

    ordenados = frutas.sort_values(
        "price usd",
        ascending=True
    )

    print(ordenados.to_string(index=False))

    print("\nNueva columna 'categoria':")

    print(
        frutas[
            ["type", "price usd", "categoria"]
        ].head(10).to_string(index=False)
    )

    print("\nCategorías utilizadas:")

    for categoria, descripcion in categorias_precio.items():
        print("-", categoria, ":", descripcion)
```

## 11. Menú principal

Finalmente se crea el menú principal del programa.

Se utiliza un ciclo `while True` para mantener el programa funcionando hasta que el usuario seleccione la opción de salir.

Las opciones del menú se recorren mediante un ciclo `for` y la función `enumerate()`.

Después se utiliza una estructura condicional `if`, `elif` y `else` para determinar qué función debe ejecutarse según la opción seleccionada.

La instrucción `break` permite finalizar el ciclo cuando el usuario selecciona la opción 9.

```python
while True:
    print("\n\n" + "=" * 55)
    print("             ANALISIS DE FRUTAS")
    print("=" * 55)

    for numero, opcion in enumerate(menu, start=1):
        print(f"{numero}. {opcion}")

    seleccion = input("\nSeleccione una opción: ")

    if seleccion == "1":
        mostrar_datos()

    elif seleccion == "2":
        consultar_fruta()

    elif seleccion == "3":
        consultar_color()

    elif seleccion == "4":
        consultar_precio()

    elif seleccion == "5":
        resumen_precios()

    elif seleccion == "6":
        analizar()

    elif seleccion == "7":
        mostrar_opciones()

    elif seleccion == "8":
        organizar_datos()

    elif seleccion == "9":
        print("\n" + "=" * 55)
        print("Gracias por utilizar el programa.")
        print("=" * 55)
        break

    else:
        print("\nOpción no válida.")
```
